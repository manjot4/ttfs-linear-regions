```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))

from ttfs_regions.utils import resolve_device
DEVICE = resolve_device("auto")
print("repo:", ROOT)
print("device:", DEVICE)
```

```python
from ttfs_regions.configs import WidthDepthConfig
from ttfs_regions.runners import run_width_depth
from ttfs_regions.plotting import plot_width_depth

DATASET = "mnist"
FAST = False
DATA_ROOT = ROOT / "data"
RESULTS_ROOT = ROOT / "results"
```

```python
cfg = (
    WidthDepthConfig(
        steps=100, seeds=(42,), num_pairs=2, calibration_size=16,
        widths=(8, 16), schedules={"8": [8, 8], "16": [16, 16]},
        shared_batch_size=32, delays_on=False,
    )
    if FAST else WidthDepthConfig()
)
run = run_width_depth(DATASET, "shared", DATA_ROOT, RESULTS_ROOT, config=cfg, device=DEVICE)
display(run["width_summary"])
plot_width_depth(run["width_summary"], run["depth_summary"], cfg.schedules,
                 RESULTS_ROOT / f"{DATASET}_shared_notebook.png")
```
