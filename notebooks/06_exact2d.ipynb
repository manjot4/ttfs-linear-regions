# Exact 2D TTFS SNN/ReLU enumeration

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))

from ttfs_regions.utils import resolve_device
DEVICE = resolve_device("auto")
print("repo:", ROOT)
print("device:", DEVICE)
```

```python
from ttfs_regions.configs import Exact2DConfig
from ttfs_regions.exact_runner import run_exact2d

DATASET = "mnist"
FAST = True
DATA_ROOT = ROOT / "data"
RESULTS_ROOT = ROOT / "results" / "exact2d"
```

```python
if FAST:
    width, depth, epochs, checkpoints = 2, 1, 0, (0,)
    cfg = Exact2DConfig(max_cells=5000, validation_grid_res=20, delays_on=False)
else:
    # Current paper figure notebook: depth 3, width 10, checkpoints 0/3/6,
    # TTFS delays enabled but frozen.
    width, depth, epochs, checkpoints = 10, 3, 10, (0, 3, 6)
    cfg = Exact2DConfig(max_cells=1_000_000, validation_grid_res=90,
                        delays_on=True, train_delays=False)

run = run_exact2d(
    DATASET, DATA_ROOT, RESULTS_ROOT,
    width=width, depth=depth, seed=42,
    epochs=epochs, checkpoints=checkpoints,
    config=cfg, device=DEVICE,
    validate_grid=True, make_plots=True,
)
display(run["results"])
print("slice bounds:", run["slice"].bounds)
print("reference labels/indices:", [(y, i) for _, y, i in run["references"]])
```

The exact count is for the bounded two-dimensional affine slice only.
