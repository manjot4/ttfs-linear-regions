# Training depth-one SNN

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))

from ttfs_regions.utils import resolve_device
DEVICE = resolve_device("auto")
print("repo:", ROOT)
print("device:", DEVICE)
```

```python
from ttfs_regions.configs import TrainingConfig
from ttfs_regions.runners import run_depth1_training
from ttfs_regions.plotting import plot_training_regions, plot_training_accuracy

DATASET = "mnist"
FAST = False
DATA_ROOT = ROOT / "data"
RESULTS_ROOT = ROOT / "results"
```

```python
if FAST:
    cfg = TrainingConfig(
        epochs=1, checkpoints=(0, 1), widths=(8,), seeds=(42,),
        region_steps=50, region_pairs=1, run_training_oriented_accuracy=True,
    )
else:
    cfg = TrainingConfig()

run = run_depth1_training(
    DATASET, DATA_ROOT, RESULTS_ROOT, config=cfg, device=DEVICE,
    widths=cfg.widths, seeds=cfg.seeds,
)
display(run["regions"].head())
display(run["accuracy"].head())
display(run["loss"].head())
plot_training_regions(run["regions"], RESULTS_ROOT / f"{DATASET}_training_regions_notebook.png")
plot_training_accuracy(run["accuracy"], RESULTS_ROOT / f"{DATASET}_training_accuracy_notebook.png")
```
